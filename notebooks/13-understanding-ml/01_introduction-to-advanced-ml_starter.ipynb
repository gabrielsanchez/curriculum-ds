# Introduction to Advanced ML Models

Understand when and why to use advanced algorithms beyond linear models and single trees.

## What You Already Know

You've mastered foundational algorithms:
- **Logistic Regression:** Linear decision boundary, fast, interpretable
- **Linear Regression:** Fit lines to continuous data, simple, reliable
- **Decision Trees:** Non-linear, interpretable, but prone to overfitting
- **Regularization (Ridge, Lasso):** Control overfitting, feature selection

These are genuinely useful — used in production systems everywhere.

## The Limitations

But these methods leave performance on the table in common scenarios:

| Scenario | Why Linear Models Struggle | Why Single Trees Struggle |
|----------|---------------------------|---------------------------|
| Complex non-linear boundaries | Can't fit curves | Overfit with deeper trees |
| Very high-dimensional data | Regularization needed | Few features per split |
| Small datasets, many features | Overfits | Sensitive to noise |
| Noisy tabular data | Underfits non-linear patterns | Memorizes noise |
| Mixed data types (cat + numeric) | Need encoding | Handles naturally |

**Solution:** Five more algorithms, each addressing different gaps.

## Five Advanced Algorithms

### 1. k-Nearest Neighbors (kNN)

**How it works:** Classify a new point by looking at its k closest neighbors and taking a majority vote.

**Strengths:**
- Simple conceptually
- Non-parametric (no assumptions about data distribution)
- Works for both classification and regression

**Weaknesses:**
- Slow at prediction time (searches entire dataset)
- Sensitive to which distance metric you use
- Features must be scaled properly
- **Curse of dimensionality:** Breaks down in very high dimensions

**When to use:**
- Small datasets (< 100k examples)
- Medium number of features (< 100)
- Want a quick baseline without much tuning

### 2. Support Vector Machines (SVM)

**How it works:** Find the optimal boundary between classes — the hyperplane with the widest "margin" (street between classes).

**Strengths:**
- Mathematically elegant (well-understood theory)
- Kernel trick allows non-linear boundaries
- Works well on moderate-sized datasets
- Memory efficient (only stores support vectors)

**Weaknesses:**
- Slower on very large datasets
- Needs feature scaling
- Hyperparameter tuning important (C, gamma)
- Less interpretable than trees

**When to use:**
- Clean, structured data
- Need to capture non-linear patterns
- Computing resources available for training

### 3. Random Forests

**How it works:** Build hundreds of independent decision trees, each on random subsets of data, then average their predictions.

**Strengths:**
- Ensemble reduces overfitting dramatically
- No feature scaling needed
- Fast training and prediction
- Feature importance built-in
- Handles non-linear relationships well

**Weaknesses:**
- Less interpretable than single tree
- Many trees = more memory
- Hyperparameter tuning (n_trees, tree depth)

**When to use:**
- General-purpose workhorse
- Large datasets (millions of rows)
- Need feature importance
- Want to stay away from tuning

### 4. Gradient Boosting

**How it works:** Build trees sequentially, each new tree corrects the mistakes of the previous ones (not independent like Random Forests).

**Strengths:**
- Often best accuracy on tabular data
- Dominates data science competitions
- Production implementations (XGBoost, LightGBM) are very fast
- Handles non-linear patterns well

**Weaknesses:**
- More hyperparameters to tune (learning rate, n_trees, depth)
- Prone to overfitting if not careful
- Slower training than Random Forests
- Less interpretable

**When to use:**
- Want highest accuracy possible
- Can afford tuning time
- Working on Kaggle competition or research

### 5. Neural Networks

**How it works:** Stack layers of neurons, each learning hierarchical features. Each neuron applies a non-linear function to weighted sum of inputs.

**Strengths:**
- Extremely flexible, can approximate any function
- State-of-the-art on images, text, audio
- Deep networks (many layers) learn hierarchical features
- Massive community, lots of libraries

**Weaknesses:**
- Needs lot of data (100k+ examples) to train well
- Slow training, requires GPU for large networks
- Hard to interpret (black box)
- Many hyperparameters, hard to tune
- Can overfit if not careful

**When to use:**
- Massive datasets (millions of examples)
- Images, text, audio data
- Need state-of-the-art accuracy
- Have time to experiment and tune

## The No Free Lunch Theorem

**Every algorithm has strengths and weaknesses. No single algorithm is best for all problems.**

Your job as a data scientist:
1. Understand your data: size, dimensionality, types of features
2. Understand your constraints: speed, interpretability, accuracy importance
3. Start simple (logistic regression)
4. Iterate, trying new algorithms
5. Use cross-validation to compare fairly
6. Pick the model with best trade-off for your use case
