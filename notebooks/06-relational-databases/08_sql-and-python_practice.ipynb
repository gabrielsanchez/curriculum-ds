## Setup: Create a test database

```python
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# Create tables
cursor.execute("""
    CREATE TABLE products (
        prod_id INTEGER PRIMARY KEY,
        name TEXT,
        price REAL,
        category TEXT
    )
""")

cursor.execute("""
    CREATE TABLE sales (
        sale_id INTEGER PRIMARY KEY,
        prod_id INTEGER,
        quantity INTEGER,
        date TEXT,
        FOREIGN KEY (prod_id) REFERENCES products(prod_id)
    )
""")

# Insert data
products = [(1, 'Laptop', 999, 'Electronics'), (2, 'Mouse', 25, 'Electronics'), (3, 'Desk', 300, 'Furniture')]
cursor.executemany("INSERT INTO products VALUES (?, ?, ?, ?)", products)

sales = [(1, 1, 2, '2024-01-10'), (2, 2, 5, '2024-01-11'), (3, 3, 1, '2024-01-12'), (4, 1, 1, '2024-01-13')]
cursor.executemany("INSERT INTO sales VALUES (?, ?, ?, ?)", sales)

conn.commit()
print("Database setup complete")
```
