# Joining Tables - Practice

Use different join types to combine tables in various scenarios.

```python
import sqlite3

conn = sqlite3.connect(":memory:")
cursor = conn.cursor()
```

## Setup: Sample data

```python
# Create employees and projects tables
cursor.execute("""
    CREATE TABLE employees (
        emp_id INTEGER PRIMARY KEY,
        name TEXT,
        department TEXT
    )
""")

cursor.execute("""
    CREATE TABLE projects (
        proj_id INTEGER PRIMARY KEY,
        name TEXT,
        emp_id INTEGER,
        FOREIGN KEY (emp_id) REFERENCES employees(emp_id)
    )
""")

# Insert data
emp_data = [(1, 'Alice', 'Engineering'), (2, 'Bob', 'Marketing'), (3, 'Charlie', 'Engineering'), (4, 'Diana', 'Sales')]
cursor.executemany("INSERT INTO employees VALUES (?, ?, ?)", emp_data)

proj_data = [(1, 'ProjectA', 1), (2, 'ProjectB', 2), (3, 'ProjectC', 1), (4, 'ProjectD', 5)]
cursor.executemany("INSERT INTO projects VALUES (?, ?, ?)", proj_data)
conn.commit()
```
