# Setting Up SQLite

Connect to and create SQLite databases using Python.

## What is SQLite?

- Lightweight, file-based database
- No server needed
- Built into Python's standard library
- Perfect for learning SQL and small projects

## The Workflow

1. **Connect** to a database
2. **Create a cursor** to execute SQL
3. **Execute** SQL statements
4. **Fetch** results or commit changes
5. **Close** the connection

## Basic Connection

```python
import sqlite3

# Create or open a database
conn = sqlite3.connect(":memory:")  # :memory: is in-memory database (disappears when script ends)
# Or use a filename: conn = sqlite3.connect("bookstore.db")

print("Connected to database")
```

## Create a Cursor

```python
# Create a cursor to execute SQL
cursor = conn.cursor()
print("Cursor created")
```

## Execute SQL: Create a Table

```python
# Create a simple customers table
cursor.execute("""
CREATE TABLE IF NOT EXISTS customers (
    customer_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    email TEXT,
    city TEXT
)
""")

conn.commit()  # Save changes
print("Table created")
```

## Insert Data

```python
# Insert sample data
cursor.execute("""
INSERT INTO customers (name, email, city)
VALUES ('Alex Carter', 'alex@email.com', 'Chicago')
""")

cursor.execute("""
INSERT INTO customers (name, email, city)
VALUES ('Jordan Lee', 'jordan@email.com', 'New York')
""")

cursor.execute("""
INSERT INTO customers (name, email, city)
VALUES ('Morgan Singh', 'morgan@email.com', 'Austin')
""")

conn.commit()  # Save changes
print("Data inserted")
```

## Query Data

```python
# Retrieve all customers
cursor.execute("SELECT * FROM customers")
rows = cursor.fetchall()

print("All customers:")
for row in rows:
    print(row)
```

## Query with Where Clause

```python
# Find customers in Chicago
cursor.execute("SELECT name, city FROM customers WHERE city = 'Chicago'")
chicago_customers = cursor.fetchall()

print("Chicago customers:")
for row in chicago_customers:
    print(row)
```

## Close the Connection

```python
# Always close when done
conn.close()
print("Connection closed")
```
