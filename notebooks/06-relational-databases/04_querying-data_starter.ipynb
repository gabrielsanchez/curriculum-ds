# Querying Data

Master SELECT queries with WHERE, DISTINCT, LIMIT, and filtering operators.

```python
import sqlite3
import pandas as pd
```

## Setup: Create Sample Database

```python
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# Create and populate customers
cursor.execute("""
CREATE TABLE customers (
    customer_id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT
)
""")

cursor.executemany("""
INSERT INTO customers VALUES (?, ?, ?)
""", [
    (1, 'Alex Carter', 'Chicago'),
    (2, 'Jordan Lee', 'New York'),
    (3, 'Morgan Singh', 'Austin'),
    (4, 'Sam Torres', 'Chicago'),
    (5, 'Taylor Reyes', 'Denver')
])

conn.commit()
print("Sample database created")
```

## SELECT All Columns

```python
cursor.execute("SELECT * FROM customers")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

## SELECT Specific Columns

```python
cursor.execute("SELECT name, city FROM customers")
rows = cursor.fetchall()
print("Name and City:")
for row in rows:
    print(row)
```

## WHERE Clause: Filter Rows

```python
# Find customers in Chicago
cursor.execute("SELECT name, city FROM customers WHERE city = 'Chicago'")
chicago = cursor.fetchall()
print("Chicago customers:")
for row in chicago:
    print(row)
```

## Multiple WHERE Conditions

```python
# Complex condition
cursor.execute("""
SELECT name, city FROM customers
WHERE city IN ('Chicago', 'Denver')
""")
rows = cursor.fetchall()
print("Customers in Chicago or Denver:")
for row in rows:
    print(row)
```

## DISTINCT: Remove Duplicates

```python
# Get unique cities
cursor.execute("SELECT DISTINCT city FROM customers")
cities = cursor.fetchall()
print("Unique cities:")
for city in cities:
    print(city[0])
```

## LIMIT: Get Top N Rows

```python
# Get first 2 customers
cursor.execute("SELECT name, city FROM customers LIMIT 2")
rows = cursor.fetchall()
print("First 2 customers:")
for row in rows:
    print(row)
```

## Load into Pandas DataFrame

```python
df = pd.read_sql_query("SELECT * FROM customers", conn)
print(df)
print(f"\nDataFrame shape: {df.shape}")
```
