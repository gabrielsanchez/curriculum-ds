# Introduction to SQL

Learn the basics of SQL and relational databases.

## What is SQL?

**SQL (Structured Query Language)** is the standard language for interacting with relational databases.

- Used to store, query, and manage data
- Works with most databases: SQLite, PostgreSQL, MySQL, Microsoft SQL Server, BigQuery
- Essential skill for data scientists

## Basic Concepts

### Table
A collection of rows and columns, like a spreadsheet.

### Row (Record)
A single entry in a table.

### Column (Field)
An attribute that all rows have (e.g., name, email, age).

### Primary Key
A unique identifier for each row (e.g., customer_id).

### Foreign Key
A column that references the primary key of another table.

## SQL Statement Types

| Category | Statements | Purpose |
|----------|-----------|----------|
| DDL | CREATE, DROP, ALTER | Define table structure |
| DML | INSERT, UPDATE, DELETE | Modify data |
| DQL | SELECT | Read data |
| DCL | GRANT, REVOKE | Manage permissions |

## Why SQL for Data Science?

1. **Data lives in databases** - Most production data is stored in SQL databases
2. **Scale** - SQL handles millions or billions of rows efficiently
3. **Expressiveness** - Complex questions in one query vs. many lines of code
4. **Universality** - Core SQL transfers across all major databases

## Basic SQL Syntax

```sql
SELECT column1, column2
FROM table_name
WHERE condition;
```

### Example
```sql
SELECT name, city
FROM customers
WHERE city = 'Chicago';
```

This query retrieves the name and city of all customers in Chicago.
