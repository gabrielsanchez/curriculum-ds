# Grouping and Sorting

Aggregate data with GROUP BY and sort results with ORDER BY.

```python
import sqlite3

conn = sqlite3.connect("bookstore.db")
cursor = conn.cursor()
```

## ORDER BY — Sorting results

```python
# Sort books by price (ascending)
cursor.execute("SELECT title, price FROM books ORDER BY price ASC")
print("Books sorted by price (cheapest first):")
for row in cursor.fetchall():
    print(f"  {row[0]} - ${row[1]}")
```

## ORDER BY DESC — Reverse order

```python
# Most expensive books first
cursor.execute("SELECT title, price FROM books ORDER BY price DESC LIMIT 3")
print("3 most expensive books:")
for row in cursor.fetchall():
    print(f"  {row[0]} - ${row[1]}")
```

## Aggregate functions

```python
# COUNT, SUM, AVG, MIN, MAX
cursor.execute("""
    SELECT
        COUNT(*) AS total_books,
        AVG(price) AS avg_price,
        MIN(price) AS cheapest,
        MAX(price) AS most_expensive
    FROM books
""")

row = cursor.fetchone()
print(f"Total books: {row[0]}")
print(f"Average price: ${row[1]:.2f}")
print(f"Cheapest: ${row[2]}")
print(f"Most expensive: ${row[3]}")
```

## GROUP BY — Aggregate by category

```python
# Count books by genre
cursor.execute("""
    SELECT genre, COUNT(*) as book_count, AVG(price) as avg_price
    FROM books
    GROUP BY genre
""")

print("Books by genre:")
for row in cursor.fetchall():
    print(f"  {row[0]}: {row[1]} books, avg price ${row[2]:.2f}")
```

## HAVING — Filter groups after aggregation

```python
# Get genres with average price > $20
cursor.execute("""
    SELECT genre, COUNT(*) as book_count, AVG(price) as avg_price
    FROM books
    GROUP BY genre
    HAVING AVG(price) > 20
""")

print("Genres with avg price > $20:")
for row in cursor.fetchall():
    print(f"  {row[0]}: {row[1]} books, avg ${row[2]:.2f}")
```

## GROUP BY with ORDER BY

```python
# Which customers placed the most orders?
cursor.execute("""
    SELECT c.name, COUNT(o.order_id) as order_count
    FROM customers c
    LEFT JOIN orders o ON c.customer_id = o.customer_id
    GROUP BY c.name
    ORDER BY order_count DESC
""")

print("Orders by customer:")
for row in cursor.fetchall():
    print(f"  {row[0]}: {row[1]} orders")
```
