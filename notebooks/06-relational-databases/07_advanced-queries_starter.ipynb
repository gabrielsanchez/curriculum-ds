# Advanced Queries

Use subqueries, CASE expressions, and transformations in SQL.

```python
import sqlite3

conn = sqlite3.connect("bookstore.db")
cursor = conn.cursor()
```

## Column aliases for clarity

```python
cursor.execute("""
    SELECT
        title,
        price,
        price * 0.9 AS discounted_price,
        stock * price AS inventory_value
    FROM books
""")

print("Books with calculated fields:")
for row in cursor.fetchall():
    print(f"  {row[0]}: ${row[1]} (sale: ${row[2]:.2f})")
```

## CASE WHEN — Conditional logic

```python
cursor.execute("""
    SELECT
        title,
        price,
        CASE
            WHEN price < 20 THEN 'Budget'
            WHEN price BETWEEN 20 AND 40 THEN 'Mid-range'
            ELSE 'Premium'
        END AS price_tier
    FROM books
    ORDER BY price
""")

print("Books by price tier:")
for row in cursor.fetchall():
    print(f"  {row[0]} (${row[1]}): {row[2]}")
```

## COALESCE — Handle NULL values

```python
cursor.execute("""
    SELECT
        COALESCE(c.city, 'Unknown') AS city,
        COUNT(*) AS customer_count
    FROM customers c
    GROUP BY c.city
""")

print("Customers by city (NULL replaced with 'Unknown'):")
for row in cursor.fetchall():
    print(f"  {row[0]}: {row[1]}")
```

## Subqueries — nested queries

```python
# Find books more expensive than the average
cursor.execute("""
    SELECT title, price
    FROM books
    WHERE price > (SELECT AVG(price) FROM books)
    ORDER BY price DESC
""")

print("Above-average priced books:")
for row in cursor.fetchall():
    print(f"  {row[0]}: ${row[1]}")
```

## String functions

```python
cursor.execute("""
    SELECT
        UPPER(title) AS title_upper,
        LENGTH(title) AS title_length,
        author
    FROM books
    WHERE LENGTH(title) > 10
    LIMIT 3
""")

print("Long book titles:")
for row in cursor.fetchall():
    print(f"  {row[0]} ({row[1]} chars) by {row[2]}")
```
