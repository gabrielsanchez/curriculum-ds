# Tables, Data Types, and Fields

Master table design with data types and constraints.

```python
import sqlite3
```

## SQLite Data Types

| Type | Description | Example |
|------|-------------|----------|
| INTEGER | Whole numbers | 42, -5, 1000 |
| REAL | Decimal numbers | 3.14, 99.95 |
| TEXT | Strings | 'Chicago', 'Alice' |
| BLOB | Binary data | (images, files) |
| NULL | No value | NULL |

## Table Constraints

- **PRIMARY KEY** — Uniquely identifies each row
- **FOREIGN KEY** — References a primary key in another table
- **NOT NULL** — Column must have a value
- **UNIQUE** — All values must be unique
- **DEFAULT** — Use a default value if not specified
- **CHECK** — Enforce a custom condition

## Example: Bookstore Database

```python
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# Create customers table
cursor.execute("""
CREATE TABLE IF NOT EXISTS customers (
    customer_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    email TEXT UNIQUE,
    city TEXT,
    join_date TEXT DEFAULT CURRENT_DATE
)
""")

# Create books table
cursor.execute("""
CREATE TABLE IF NOT EXISTS books (
    book_id INTEGER PRIMARY KEY,
    title TEXT NOT NULL,
    author TEXT,
    price REAL CHECK(price > 0),
    stock INTEGER DEFAULT 0
)
""")

# Create orders table with foreign keys
cursor.execute("""
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER NOT NULL,
    book_id INTEGER NOT NULL,
    order_date TEXT,
    quantity INTEGER CHECK(quantity > 0),
    FOREIGN KEY(customer_id) REFERENCES customers(customer_id),
    FOREIGN KEY(book_id) REFERENCES books(book_id)
)
""")

conn.commit()
print("Tables created successfully")
```

## Insert Sample Data

```python
# Insert customers
cursor.execute("""
INSERT INTO customers (name, email, city) VALUES
('Alex Carter', 'alex@email.com', 'Chicago'),
('Jordan Lee', 'jordan@email.com', 'New York'),
('Morgan Singh', 'morgan@email.com', 'Austin')
""")

# Insert books
cursor.execute("""
INSERT INTO books (title, author, price, stock) VALUES
('Python Crash Course', 'Eric Matthes', 35.99, 10),
('Data Science Handbook', 'Jake VanderPlas', 45.99, 15),
('SQL for Beginners', 'Unknown', 24.99, 20)
""")

# Insert orders
cursor.execute("""
INSERT INTO orders (customer_id, book_id, order_date, quantity) VALUES
(1, 1, '2024-01-15', 1),
(2, 2, '2024-01-20', 2),
(3, 3, '2024-01-25', 1)
""")

conn.commit()
print("Data inserted successfully")
```

## Query the Data

```python
# Show customers
print("Customers:")
cursor.execute("SELECT * FROM customers")
for row in cursor.fetchall():
    print(row)

print("\nBooks:")
cursor.execute("SELECT * FROM books")
for row in cursor.fetchall():
    print(row)

print("\nOrders:")
cursor.execute("SELECT * FROM orders")
for row in cursor.fetchall():
    print(row)
```
