# Supervised vs. Unsupervised Learning

Distinguish learning tasks based on whether data is labeled.

## The Key Question: Do You Have Labels?

The most important decision in ML is: **Does your training data have correct answers (labels)?**

| Supervised | Unsupervised |
|------------|-------------|
| ✅ Data has labels (correct answers) | ❌ Data has no labels |
| Learn from examples with known outputs | Discover structure/patterns yourself |
| Require more effort to label data | Useful when labeling is expensive/impossible |
| Examples: emails labeled spam/not-spam | Examples: grouping customers into segments |

## Supervised Learning

Training data format:
```
Features (X)               Label (y)
[square_ft, bedrooms, ...] → Price ($)
[age, income, credit_...] → Loan approved (Yes/No)
[word1, word2, word3, ...] → Spam (Yes/No)
```

The model learns a function: `f(X) ≈ y`

### Two types of supervised learning:

### 1. Regression: Predict a Number

**Target:** Continuous numeric value

**Examples:**
- Predict house price ($100,000 - $1,000,000)
- Forecast temperature (60°F - 90°F)
- Estimate delivery time (15 - 120 minutes)
- Predict stock price ($10.50 - $500.75)

**How to recognize it:** The question is "how much?" or "how many?"

**Evaluation metric:** Mean Absolute Error, R² score

### 2. Classification: Predict a Category

**Target:** Discrete category (class label)

**Binary Classification (2 classes):**
- Email: Spam or Not Spam
- Transaction: Fraudulent or Legitimate
- Patient: Disease or Healthy
- Loan: Approved or Denied

**Multi-class Classification (3+ classes):**
- Digit recognition: 0, 1, 2, ..., 9 (10 classes)
- Sentiment: Positive, Neutral, Negative (3 classes)
- Animal: Cat, Dog, Bird, Fish (4 classes)
- Severity: Mild, Moderate, Severe (3 classes)

**How to recognize it:** The question is "which group?" or "yes/no?"

**Evaluation metric:** Accuracy, Precision, Recall, F1 Score

## Unsupervised Learning

Training data format:
```
Features (X) only - NO labels
[customer_id, purchase_history, ...]
[sentence_1, sentence_2, ...]
[product_features]
```

The model finds structure without guidance.

### 1. Clustering: Group Similar Data

**Task:** Partition data into similar groups

**Examples:**
- Customer segmentation: High-value, Medium-value, Low-value
- Document clustering: Group news articles by topic
- Gene sequencing: Group similar genetic patterns
- Recommendation: "Others like you also enjoyed..."

**Key:** You don't know the groups in advance. The algorithm discovers them.

### 2. Dimensionality Reduction: Simplify Data

**Task:** Reduce number of features while keeping information

**Why?**
- Too many features slow down models
- Some features are redundant
- Visualization: Reduce 100 features → 2D plot

**Examples:**
- Image compression: Reduce 1000 pixels → 50 compressed features
- Feature reduction: 100 health metrics → 10 key indicators
- Visualization: 50 variables → 2 principal components → plot
