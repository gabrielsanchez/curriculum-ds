# Introduction to Machine Learning

Learn what machine learning is and how it differs from traditional programming.

## What is Machine Learning?

**Machine Learning (ML):** A branch of AI where systems learn patterns from data and make predictions — without being explicitly programmed with rules.

Instead of:
```
Traditional Programming: Data + Rules → Output
```

Machine Learning follows:
```
Machine Learning: Data + Output → Rules (the model)
```

## Traditional Programming vs. Machine Learning

### Traditional Approach (Rule-Based)
A programmer writes rules explicitly.

**Example: Spam Filter**
```python
if "free money" in email.subject or "click here" in email.body:
    mark_as_spam()
else:
    allow_email()
```

Problem: Spammers evolve tactics. You'd need to constantly update rules.

### Machine Learning Approach
The model learns patterns from thousands of labeled emails.

The algorithm discovers:
- Which word combinations predict spam (without being told)
- Which sender addresses are suspicious
- How spam patterns evolve over time

Then automatically predicts on new emails.

## Key ML Concepts

| Term | Definition |
|------|------------|
| **Model** | The learned mathematical function after training |
| **Features (X)** | Input variables the model uses to make predictions |
| **Label/Target (y)** | The output variable we want to predict |
| **Training** | Process of learning the model from labeled data |
| **Prediction** | Using the trained model on new, unseen data |
| **Algorithm** | The mathematical procedure to learn the model |
| **Generalization** | Model's ability to perform well on new data |
| **Overfitting** | Model memorizes training data instead of learning patterns |

## Why ML?

When do you use machine learning instead of traditional programming?

**Use ML when:**
- Rules are too complex or numerous to code manually
- You have lots of labeled data but no clear rules
- Patterns change over time (you can retrain)
- The problem requires human-like judgment

**Examples:**
- Image recognition (too many visual patterns to code)
- Recommendation systems (relationships are complex)
- Fraud detection (tactics evolve constantly)
- Natural language processing (language is nuanced)

**Avoid ML if:**
- The problem has simple, unchanging rules
- You don't have enough labeled data
- You need explanations for every prediction
- Real-time predictions aren't needed

## The Machine Learning Workflow

Every ML project follows this sequence:

```
1. Define the problem
        ↓
2. Collect data
        ↓
3. Explore & clean data (EDA)
        ↓
4. Engineer features
        ↓
5. Select & train a model
        ↓
6. Evaluate on test data
        ↓
7. Deploy & monitor
```

**Note:** This is iterative. You often cycle back to earlier steps based on results.

## Real-World Example: House Price Prediction

**Problem:** Predict house sale prices given features

**Features (what we know):**
- Square footage
- Number of bedrooms
- Location (neighborhood)
- Age of house
- Garage spaces

**Target (what we predict):**
- Sale price ($)

**Traditional approach:**
Try to write a formula: `Price = (sqrt_ft × 150) + (bedrooms × 20000) + ...` ← Hard to get right!

**ML approach:**
Train a model on thousands of past sales. Model learns the actual relationships automatically.
