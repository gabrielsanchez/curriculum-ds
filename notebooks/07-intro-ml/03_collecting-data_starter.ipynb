# Collecting Data

Explore data sources and load public datasets in Python.

## Where Does Data Come From?

The saying: **"Garbage in, garbage out"** — ML is only as good as the data.

## Data Sources

### 1. Public Datasets (Easiest to Start)

**Repositories:**
- [Kaggle Datasets](https://www.kaggle.com/datasets) — Community datasets, competitions
- [UCI ML Repository](https://archive.ics.uci.edu/ml/) — Classic datasets (Iris, Wine, etc.)
- [Google Dataset Search](https://datasetsearch.research.google.com/)

**Government Data:**
- [data.gov](https://www.data.gov) — U.S. government data
- [World Bank Open Data](https://data.worldbank.org/) — Global data
- [CDC / NIH](https://www.cdc.gov/) — Health data

**Advantage:** Ready to use, free, no collection effort

**Disadvantage:** May not fit your exact problem

### 2. Built-in Library Datasets

```python
# scikit-learn comes with several classic datasets
from sklearn.datasets import load_iris, load_wine, load_breast_cancer

# Load as DataFrame
iris = load_iris(as_frame=True)
df = iris.frame

print(f"Shape: {df.shape}")
print(f"\nFirst few rows:")
print(df.head())
print(f"\nTarget variable (label):")
print(iris.target_names)
```

```python
# Seaborn also has built-in datasets
import seaborn as sns

titanic = sns.load_dataset('titanic')
print(f"Titanic shape: {titanic.shape}")
print(f"\nColumns: {list(titanic.columns)}")
print(f"\nFirst rows:")
print(titanic.head())
```

### 3. APIs and Web Services

Access real-time or updated data programmatically.

**Examples:**
- Weather data: OpenWeatherMap, NOAA
- Stock prices: Alpha Vantage, Yahoo Finance
- Social media: Twitter/X API, Reddit API
- Geographic: Google Maps, OpenStreetMap

**Advantage:** Fresh, real-time data

**Disadvantage:** May require authentication, rate limits, data processing
