```python
import pandas as pd
```

## Creating a DataFrame from a Dictionary

```python
data = {
    "product":  ["Laptop", "Phone", "Tablet", "Monitor"],
    "price":    [999, 699, 499, 349],
    "in_stock": [True, True, False, True]
}

df = pd.DataFrame(data)
print(df)
```

## Creating a DataFrame from a List of Dictionaries

```python
records = [
    {"name": "Alex",   "score": 88, "grade": "B"},
    {"name": "Jordan", "score": 95, "grade": "A"},
    {"name": "Morgan", "score": 72, "grade": "C"},
]

df = pd.DataFrame(records)
print(df)
```

## Setting a Custom Index

```python
records = [
    {"name": "Alex",   "score": 88, "grade": "B"},
    {"name": "Jordan", "score": 95, "grade": "A"},
    {"name": "Morgan", "score": 72, "grade": "C"},
]

df = pd.DataFrame(records)
df = df.set_index("name")
print(df)
```

## Writing to a CSV File

```python
data = {
    "product":  ["Laptop", "Phone", "Tablet"],
    "price":    [999, 699, 499],
    "quantity": [5, 12, 8]
}

df = pd.DataFrame(data)

# Save to CSV (don't include the index)
df.to_csv("products.csv", index=False)
print("File saved as products.csv")
```

## Reading from a CSV File

```python
# Read the CSV file
df_loaded = pd.read_csv("products.csv")
print(df_loaded)

# Check the info
print(df_loaded.info())
```
