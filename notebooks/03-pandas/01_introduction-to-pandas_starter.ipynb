## Installing and Importing Pandas

```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
```

## Creating a Series

A Series is a one-dimensional labeled array.

```python
# Series from a list
temperatures = pd.Series([72, 75, 68, 80, 77])
print(temperatures)

# Series with custom index
scores = pd.Series(
    [88, 92, 75, 95],
    index=["Alex", "Jordan", "Morgan", "Sam"]
)
print(scores)
```

## Accessing Series Values

```python
scores = pd.Series([88, 92, 75, 95], index=["Alex", "Jordan", "Morgan", "Sam"])

# Access by label
print(scores["Alex"])

# Access by position
print(scores[0])

# Filter values greater than 85
print(scores[scores > 85])
```

## Creating a DataFrame from a Dictionary

```python
data = {
    "name":   ["Alex", "Jordan", "Morgan", "Sam", "Taylor"],
    "age":    [25, 30, 22, 28, 35],
    "city":   ["Chicago", "New York", "Austin", "Denver", "Seattle"],
    "score":  [88, 92, 75, 95, 81]
}

df = pd.DataFrame(data)
print(df)
```

## Exploring a DataFrame

```python
# First 3 rows
print(df.head(3))

# Last 2 rows
print(df.tail(2))

# Shape: (rows, columns)
print("Shape:", df.shape)

# Data types
print(df.dtypes)

# Summary statistics
print(df.describe())
```
