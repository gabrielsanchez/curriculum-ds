## Setting Up Sample Data

```python
import pandas as pd

df = pd.DataFrame({
    "name":   ["Alex", "Jordan", "Morgan", "Sam", "Taylor"],
    "age":    [25, 30, 22, 28, 35],
    "city":   ["Chicago", "New York", "Austin", "Denver", "Seattle"],
    "score":  [88, 92, 75, 95, 81],
    "passed": [True, True, False, True, True]
})

print(df)
```

## Selecting Columns

```python
# Single column (returns a Series)
print(df["name"])

# Multiple columns (returns a DataFrame)
print(df[["name", "score"]])
```

## Selecting Rows with `loc` (label-based)

```python
# Select a single row by index label
print(df.loc[2])

# Select a range of rows (inclusive on both ends)
print(df.loc[1:3])

# Select specific rows and columns
print(df.loc[0:2, ["name", "score"]])
```

## Selecting Rows with `iloc` (position-based)

```python
# Select the first row
print(df.iloc[0])

# Select rows 1 to 3 (exclusive end)
print(df.iloc[1:3])

# Select rows 0-2, columns 0 and 2
print(df.iloc[0:3, [0, 2]])
```

## Boolean Indexing (Filtering)

```python
# Keep only rows where score is greater than 85
high_scorers = df[df["score"] > 85]
print(high_scorers)

# Multiple conditions with & (and) and | (or)
filtered = df[(df["score"] > 80) & (df["passed"] == True)]
print(filtered)
```

## Assigning New or Modified Columns

```python
# Add a new column
df["age_group"] = df["age"].apply(lambda x: "Young" if x < 30 else "Senior")
print(df)

# Modify an existing column
df["score_scaled"] = df["score"] / 100
print(df)
```
