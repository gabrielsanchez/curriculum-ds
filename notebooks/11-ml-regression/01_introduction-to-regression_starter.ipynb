# Introduction to Regression

Understand what regression is and when to use it.

## Regression vs. Classification

| Question | Answer Type | Task |
|----------|-------------|------|
| Will this customer churn? | Yes / No | **Classification** |
| How much revenue will we earn? | $412,000 | **Regression** |
| What species is this flower? | Rose / Tulip / Daisy | **Classification** |
| How long will this patient stay? | 4.2 days | **Regression** |
| Is this email spam? | Spam / Not Spam | **Classification** |
| What's this apartment's rent? | $2,200/month | **Regression** |

**Key difference:**
- **Classification:** Predict a **discrete category** from a fixed set
- **Regression:** Predict a **continuous number** that can take any value in a range

## What is Regression?

**Regression:** A supervised learning task where the goal is to predict a continuous numeric value.

Instead of "which category?", you're asking "how much?" or "how many?"

**Examples of regression targets:**
- House price (in dollars)
- Customer lifetime value (in dollars)
- Patient survival time (in months)
- Electricity demand (in kilowatts)
- Stock price (in dollars)
- Temperature (in Celsius)
- Delivery time (in minutes)

## Key Concepts

### Target Variable
The number you want to predict — continuous, can be positive, negative, fractional.

Examples:
- $250,000 (house price)
- 3.7 (customer rating out of 5)
- -2.5°C (temperature)
- 47.82 (efficiency score)

### Residuals
The errors the model makes.

```
Residual = Actual Value − Predicted Value
```

Example:
- True house price: $300,000
- Model's prediction: $295,000
- Residual: $5,000 (model underestimated)

A good model has residuals close to zero.

### Underfitting vs. Overfitting

| Underfitting | Good Fit | Overfitting |
|--------------|----------|-------------|
| Model too simple | Model captures true pattern | Model memorizes noise |
| Poor training performance | Good training performance | Excellent training performance |
| Poor test performance | Good test performance | *Poor test performance* |
| Example: Fitting a line to curved data | Example: Right complexity | Example: Fitting to every random bump |

**Goal:** Find the sweet spot in the middle.

## The Regression Pipeline

```
1. Define the Problem
   └─ What number are we predicting?
        ↓
2. Collect & Explore Data
   └─ EDA: distributions, correlations, outliers
        ↓
3. Clean & Engineer Features
   └─ Handle missing values, encode categoricals, scale numerics
        ↓
4. Train/Test Split
   └─ 80% train, 20% test
        ↓
5. Train a Regression Model
   └─ Choose algorithm (linear, polynomial, tree, etc.)
        ↓
6. Evaluate Predictions
   └─ Use MAE, RMSE, R² metrics
        ↓
7. Iterate & Improve
   └─ Try different algorithms, hyperparameters, features
```
