# Word Embeddings and Transformer Models

This notebook covers:
- Static word embeddings (Word2Vec, GloVe)
- Word analogy tasks demonstrating embedding geometry
- Contextual embeddings and self-attention
- BERT and DistilBERT models
- Fine-tuning Transformers for downstream tasks
- Hugging Face transformers library

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

print("EMBEDDINGS AND TRANSFORMERS OVERVIEW\n")
print("="*60)

# Concept: Static vs Contextual Embeddings
embedding_comparison = pd.DataFrame({
    'Aspect': ['Vector', 'Context-awareness', 'Training', 'Use case', 'Example'],
    'Static (Word2Vec)': ['Fixed per word', 'No - same for all contexts', 'Unsupervised', 'Similarity, analogies', 'king→woman=queen'],
    'Contextual (BERT)': ['Changes with context', 'Yes - varies by surrounding text', 'Pre-trained + fine-tune', 'Classification, NER', 'bank in "river bank" ≠ "investment bank"']
})

print(embedding_comparison.to_string(index=False))

print("\n\nKEY CONCEPTS:")
print("\n1. WORD EMBEDDINGS")
print("   • Represent words as dense vectors (typically 50-300 dimensions)")
print("   • Similar words have similar vectors → cosine similarity")
print("   • Trained unsupervised from large text corpora")
print("   • Examples: Word2Vec, GloVe, FastText")

print("\n2. SELF-ATTENTION (Transformers)")
print("   • Each word attends to every other word in sequence")
print("   • Computes weighted average of context")
print("   • Allows parallel processing unlike RNNs")
print("   • Foundation of BERT, GPT, T5, etc.")

print("\n3. BERT (& DistilBERT)")
print("   • Bidirectional: reads text left-to-right AND right-to-left")
print("   • Pre-trained on 3.3B words (masked language modeling)")
print("   • Fine-tune on downstream tasks with minimal additional data")
print("   • DistilBERT: 40% smaller, 60% faster, 97% of BERT performance")
```

```python
print("\n" + "="*60)
print("PRACTICAL EXAMPLE: Hugging Face Pipeline")
print("="*60)

# Note: This requires transformers library installed
example_code = """
from transformers import pipeline

# Create a text classification pipeline
classifier = pipeline("text-classification", model="distilbert-base-uncased-finetuned-sst-2-english")

texts = [
    "This movie is amazing!",
    "I didn't like this product at all.",
    "The weather is nice today."
]

results = classifier(texts)
for text, result in zip(texts, results):
    print(f"'{text}'")
    print(f"  → {result['label']}: {result['score']:.4f}\\n")
"""

print(example_code)

print("\nCommon Pre-trained Models on Hugging Face:")
models = {
    "distilbert-base-uncased": "General English, fast & lightweight",
    "bert-base-multilingual-cased": "Multilingual support",
    "PubMedBERT": "Medical/biomedical text",
    "LegalBERT": "Legal documents",
    "DistilBERT": "Sentiment analysis (SST-2 fine-tuned)"
}

for model, description in models.items():
    print(f"\n• {model}")
    print(f"  → {description}")
```
