# Sentiment Analysis Introduction

This notebook covers:
- Sentiment analysis as a core NLP task
- Positive/negative/neutral classification
- Rule-based approaches (lexicon-based)
- Machine learning approaches (TF-IDF + classifier)
- Pre-trained sentiment models
- Visualizing sentiment distributions
- Real-world datasets (movie reviews, product ratings)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("SENTIMENT ANALYSIS OVERVIEW\n")
print("="*60)

# Example reviews with sentiments
reviews = [
    ("The movie was excellent and entertaining!", "positive"),
    ("Terrible quality, complete waste of money.", "negative"),
    ("It was okay, nothing special but not bad.", "neutral"),
    ("Amazing customer service and fast delivery!", "positive"),
    ("Broke after one week, very disappointed.", "negative"),
    ("The product works as described.", "neutral"),
]

print("Example Reviews with Sentiment Labels:\n")
for review, sentiment in reviews:
    print(f"[{sentiment:8s}] {review}")

print("\n" + "="*60)
print("SENTIMENT ANALYSIS APPROACHES")
print("="*60)

approaches = pd.DataFrame({
    'Approach': ['Lexicon-based', 'ML (TF-IDF + LR)', 'Pre-trained (BERT)', 'Fine-tuned BERT'],
    'Speed': ['Fast', 'Fast', 'Medium', 'Medium'],
    'Accuracy': ['Low-Medium', 'Medium', 'High', 'Very High'],
    'Training': ['None', 'Quick', 'None', 'Hours to days'],
    'Best for': ['Baseline', 'Simple tasks', 'General sentiment', 'Domain-specific']
})

print(approaches.to_string(index=False))

print("\n\nCHALLENGES IN SENTIMENT ANALYSIS:\n")
challenges = {
    "Negation": '"NOT bad" is positive, but contains "bad"',
    "Sarcasm": '"Yeah, sure, great..."  is actually negative',
    "Context": '"The food was cold" is negative for hot dishes but maybe positive for cold desserts',
    "Aspect": '"The acting was great but the plot was boring" - mixed',
    "Domain": 'Model trained on movie reviews may not work on product reviews'
}

for challenge, example in challenges.items():
    print(f"• {challenge}: {example}")
```
