## What is NLP?

**Natural Language Processing (NLP)** enables computers to understand, interpret, and generate human language. Applications include:
- **Text Classification**: Spam detection, topic categorization, sentiment analysis
- **Named Entity Recognition (NER)**: Identifying people, places, organizations, dates, amounts
- **Machine Translation**: Translating text between languages
- **Text Summarization**: Generating concise summaries of documents
- **Question Answering**: Finding answers within documents
- **Text Generation**: Generating coherent new text (emails, code, articles)

```python
# Core NLP Concepts

# 1. TOKENS: Basic units of text
sample_text = "NLP is fascinating! It powers chatbots, search engines, and more."
tokens = sample_text.split()
print(f"Text: {sample_text}")
print(f"Tokens (simple split): {tokens}")
print(f"Number of tokens: {len(tokens)}\n")

# 2. More sophisticated tokenization (sentences)
from nltk.tokenize import sent_tokenize, word_tokenize
sentences = sent_tokenize(sample_text)
print(f"Sentences: {sentences}")

# 3. Word tokenization handles punctuation better
words = word_tokenize(sample_text)
print(f"\nWord tokens (NLTK): {words}\n")

# 4. KEY CHALLENGE: Ambiguity
ambiguous_examples = [
    "I saw her duck.",           # Is duck a verb or noun?
    "The bank is by the river",  # Bank = financial institution or riverbank?
    "I went to the bank.",       # Different meaning of "bank"
]

print("Lexical Ambiguity Examples:")
for example in ambiguous_examples:
    print(f"  '{example}' - Multiple interpretations possible")
```

```python
# The NLP Pipeline
print("="*60)
print("NLP PIPELINE OVERVIEW")
print("="*60)

pipeline = {
    "1. Text Acquisition": "Collect raw text (documents, reviews, tweets, etc.)",
    "2. Preprocessing": "Tokenization, lowercasing, stopword removal, stemming",
    "3. Feature Extraction": "Convert text to numerical representation (TF-IDF, embeddings)",
    "4. Model": "Apply ML/DL model (classification, NER, generation)",
    "5. Post-processing": "Format output, extract entities, generate text",
    "6. Evaluation": "Measure accuracy, precision, recall, F1-score"
}

for step, description in pipeline.items():
    print(f"\n{step}")
    print(f"  → {description}")

print("\n" + "="*60)
print("KEY LIBRARIES FOR NLP")
print("="*60)
libraries = {
    "NLTK": "Classic NLP toolkit: tokenization, POS tagging, stemming",
    "spaCy": "Production-grade NLP: fast, integrated pipelines, NER",
    "scikit-learn": "Text feature extraction: TF-IDF, CountVectorizer",
    "Hugging Face": "Pre-trained Transformers: BERT, GPT, T5, fine-tuning",
    "Gensim": "Word embeddings: Word2Vec, GloVe, Doc2Vec"
}

for lib, purpose in libraries.items():
    print(f"\n{lib}")
    print(f"  → {purpose}")
```
