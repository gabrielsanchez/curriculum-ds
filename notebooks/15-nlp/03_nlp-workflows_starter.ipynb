# NLP Workflows: End-to-End Pipelines

This notebook covers:
- Complete text classification workflow
- Named Entity Recognition (NER)
- Text summarization
- Zero-shot classification
- Choosing between classical vs. Transformer methods
- Decision frameworks for NLP tasks

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
import numpy as np

# Example: Customer Support Ticket Classification
tickets = [
    ("My order hasn't arrived yet!", "shipping"),
    ("I was charged twice!", "billing"),
    ("The product is damaged.", "quality"),
    ("When will my package arrive?", "shipping"),
    ("My refund hasn't processed.", "billing"),
    ("This item broke after one week.", "quality"),
]

X = [text for text, _ in tickets]
y = [label for _, label in tickets]

print("WORKFLOW 1: TEXT CLASSIFICATION\n")
print("Dataset:")
for text, label in tickets:
    print(f"  [{label:10s}] {text}")

# Create simple classification pipeline
pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(max_features=50, stop_words='english')),
    ('classifier', LogisticRegression(max_iter=200))
])

pipeline.fit(X, y)

# Test predictions
test_texts = [
    "Where is my delivery?",
    "I was billed incorrectly.",
    "It stopped working!"
]

print("\n" + "="*60)
print("Predictions on New Tickets:")
print("="*60)
predictions = pipeline.predict(test_texts)
probabilities = pipeline.predict_proba(test_texts)

for text, pred in zip(test_texts, predictions):
    print(f"\n'{text}'")
    print(f"  → Predicted category: {pred}")
```

```python
# Workflow decision framework
print("\n" + "="*60)
print("CHOOSING THE RIGHT METHOD")
print("="*60)

decision_tree = """
TEXT CLASSIFICATION DECISION TREE:

Do you have labeled training data?
├── No  → Use zero-shot classification (Hugging Face pipeline)
│         • No training needed
│         • Fast but less accurate
│
└── Yes → How many labeled examples?
          ├── < 500 examples
          │   → Fine-tune DistilBERT (few-shot learning)
          │   • Use transfer learning
          │   • High accuracy with small data
          │
          ├── 500–5,000 examples
          │   → Compare TF-IDF + LogisticRegression vs DistilBERT
          │   • TF-IDF: Fast, interpretable, often sufficient
          │   • DistilBERT: Better accuracy, slower training
          │
          └── > 5,000 examples
              → TF-IDF + LR often sufficient
              • Fine-tune BERT if accuracy is critical
              • Trade-off: speed vs marginal accuracy gains

Other considerations:
• Domain-specific text? → Use domain-specific BERT variant
  (PubMedBERT for medical, LegalBERT for legal, CodeBERT for code)
• Multilingual? → Use mBERT (multilingual BERT)
• Real-time predictions required? → Use lighter models (DistilBERT)
"""

print(decision_tree)
```
