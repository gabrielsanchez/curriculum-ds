```python
import numpy as np

print("="*60)
print("EXERCISE: Understanding Embeddings")
print("="*60)

# Simulate word embeddings conceptually
# In practice you'd use Word2Vec, GloVe, or BERT

# Example: Creating synthetic embeddings to demonstrate concepts
np.random.seed(42)

words = ["king", "queen", "man", "woman", "prince", "princess"]
n_features = 50

# Create synthetic embeddings (in practice: trained from data)
embeddings = {
    "king": np.random.randn(n_features),
    "queen": np.random.randn(n_features),
    "man": np.random.randn(n_features),
    "woman": np.random.randn(n_features),
    "prince": np.random.randn(n_features),
    "princess": np.random.randn(n_features),
}

# Simulate embedding space where:
# - king and queen are similar (royalty)
# - man and woman are similar (gender)
# - Relationship: king - man ≈ queen - woman

embeddings["king"] = np.concatenate([np.ones(10), np.ones(10), np.random.randn(30)])
embeddings["queen"] = np.concatenate([np.ones(10), -np.ones(10), np.random.randn(30)])
embeddings["man"] = np.concatenate([-np.ones(10), np.ones(10), np.random.randn(30)])
embeddings["woman"] = np.concatenate([-np.ones(10), -np.ones(10), np.random.randn(30)])

# Normalize
for word in embeddings:
    embeddings[word] = embeddings[word] / np.linalg.norm(embeddings[word])

def cosine_similarity(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

print("\nWord Pair Similarities:")
print(f"{'Pair':<20} {'Cosine Similarity':>18}")
print("-"*40)

pairs = [("king", "queen"), ("man", "woman"), ("king", "man"), ("king", "princess")]
for w1, w2 in pairs:
    sim = cosine_similarity(embeddings[w1], embeddings[w2])
    print(f"{w1} ↔ {w2:<13} {sim:>18.4f}")

print("\n\nWord Analogy (demonstrating embedding geometry):")
print("Word relationships are encoded as vectors!")
print("\nAnalogy: king - man + woman = ?")

# Approximate analogy
king_vec = embeddings["king"]
man_vec = embeddings["man"]
woman_vec = embeddings["woman"]

# Compute: king - man + woman
analogy_vec = king_vec - man_vec + woman_vec
analogy_vec = analogy_vec / np.linalg.norm(analogy_vec)

# Find closest word
similarities = {word: cosine_similarity(analogy_vec, embeddings[word]) for word in embeddings}
closest = max(similarities, key=similarities.get)

print(f"Result should be 'queen': closest word is '{closest}'")
print(f"Similarity to queen: {similarities['queen']:.4f}")
```
