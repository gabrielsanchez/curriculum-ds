## Why Visualization Matters

Visualization helps you:
- **Spot outliers** — see data points far outside the normal range
- **Understand distributions** — see if data is normally distributed, skewed, or multimodal
- **Identify relationships** — see if variables move together
- **Detect trends** — see patterns over time that are invisible in tables

## Chart Types and When to Use Them

### Line Chart
**Best for:** Trends over time or ordered categories.
- x-axis is usually time or an ordered sequence
- y-axis shows the measured value
- Great for stock prices, monthly sales, website traffic

### Bar Chart
**Best for:** Comparing values across discrete categories.
- Use for revenue, population, or counts across different groups
- Grouped bars compare multiple series
- Stacked bars show part-to-whole relationships

### Scatter Plot
**Best for:** Exploring relationships between two continuous variables.
- Shows whether variables are correlated
- Useful for spotting clusters and outliers
- Basis of correlation analysis and regression

### Histogram
**Best for:** Showing the distribution of a single continuous variable.
- Shows how values are spread out
- Reveals if data is symmetric, skewed, or bimodal
- Bars touch each other (unlike bar charts)

### Box Plot
**Best for:** Comparing distributions across groups and spotting outliers.
- Shows median, quartiles, and outliers
- Great for side-by-side distribution comparison

## Principles of Effective Visualization

1. **Choose the right chart type** — Different data questions need different charts
2. **Use meaningful titles and labels** — Make it clear what you're showing
3. **Avoid clutter** — Remove unnecessary elements
4. **Use color thoughtfully** — Color should highlight, not confuse
5. **Provide context** — Show axis labels, units, and scales
