## Installing and Importing Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

print("Matplotlib version:", plt.matplotlib.__version__)
```

## Creating a Scatter Plot

```python
# Create sample data
df_scatter = pd.DataFrame({
    'X': np.random.rand(30),
    'Y': np.random.rand(30),
    'Label': np.random.choice(['Type A', 'Type B', 'Type C'], 30)
})

# Create scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(df_scatter['X'], df_scatter['Y'])
plt.title("Scatter Plot Example")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## Creating a Line Chart

```python
# Create sample data
df_line = pd.DataFrame({
    'Year': range(2015, 2024),
    'Sales': [150, 180, 165, 200, 220, 210, 240, 250, 280]
})

# Create line chart
plt.figure(figsize=(10, 6))
plt.plot(df_line['Year'], df_line['Sales'], marker='o', linewidth=2, color='blue')
plt.title("Line Chart Example")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True, alpha=0.3)
plt.show()
```

## Creating a Bar Chart

```python
# Create sample data
df_bar = pd.DataFrame({
    'Category': ['A', 'B', 'C', 'D'],
    'Value': [23, 45, 56, 78]
})

# Create bar chart
plt.figure(figsize=(8, 6))
plt.bar(df_bar['Category'], df_bar['Value'], color='skyblue', edgecolor='navy')
plt.title("Bar Chart Example")
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()
```

## Creating a Histogram

```python
# Create sample data
data = np.random.normal(100, 15, 1000)

# Create histogram
plt.figure(figsize=(8, 6))
plt.hist(data, bins=30, color='green', alpha=0.7, edgecolor='black')
plt.title("Histogram Example")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Customizing with Titles, Labels, and Legends

```python
# Create multiple series
years = range(2015, 2024)
sales_a = [100, 120, 140, 135, 155, 170, 180, 185, 200]
sales_b = [80, 95, 110, 120, 130, 140, 150, 160, 175]

# Create figure with custom styling
plt.figure(figsize=(10, 6))
plt.plot(years, sales_a, marker='o', label='Product A', linewidth=2)
plt.plot(years, sales_b, marker='s', label='Product B', linewidth=2)
plt.title("Sales Comparison Over Time", fontsize=14, fontweight='bold')
plt.xlabel("Year", fontsize=12)
plt.ylabel("Sales ($1000s)", fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
