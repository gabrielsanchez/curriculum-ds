## Installing and Importing Plotly

```python
import plotly.express as px
import numpy as np
import pandas as pd

print("Plotly is ready!")
```

## Creating a Scatter Plot

```python
# Create sample data
df_scatter = pd.DataFrame({
    'X': np.random.rand(30),
    'Y': np.random.rand(30),
    'Label': np.random.choice(['Type A', 'Type B', 'Type C'], 30)
})

# Create interactive scatter plot
fig = px.scatter(
    df_scatter,
    x='X',
    y='Y',
    color='Label',
    title="Interactive Scatter Plot"
)
fig.show()
```

## Creating a Line Chart

```python
# Create sample data
df_line = pd.DataFrame({
    'Year': range(2015, 2024),
    'Sales': [150, 180, 165, 200, 220, 210, 240, 250, 280]
})

# Create interactive line chart
fig = px.line(
    df_line,
    x='Year',
    y='Sales',
    title="Interactive Line Chart",
    markers=True
)
fig.show()
```

## Creating a Bar Chart

```python
# Create sample data
df_bar = pd.DataFrame({
    'Category': ['A', 'B', 'C', 'D', 'E'],
    'Value': [23, 45, 56, 78, 32]
})

# Create interactive bar chart
fig = px.bar(
    df_bar,
    x='Category',
    y='Value',
    title="Interactive Bar Chart",
    color='Value'
)
fig.show()
```

## Creating a Grouped Bar Chart

```python
# Create sample data
df_grouped = pd.DataFrame({
    'Group': ['Q1', 'Q1', 'Q2', 'Q2', 'Q3', 'Q3', 'Q4', 'Q4'],
    'Region': ['North', 'South', 'North', 'South', 'North', 'South', 'North', 'South'],
    'Sales': [150, 120, 180, 140, 200, 160, 220, 180]
})

# Create interactive grouped bar chart
fig = px.bar(
    df_grouped,
    x='Group',
    y='Sales',
    color='Region',
    barmode='group',
    title="Grouped Bar Chart Comparison"
)
fig.show()
```

## Key Features of Plotly

- **Hover tooltips** — See exact values when you hover over data points
- **Zoom and pan** — Click and drag to explore specific areas
- **Toggle legend items** — Click legend entries to show/hide data series
- **Download** — Save the chart as a PNG image
- **Interactive legend** — Double-click to isolate a single series
