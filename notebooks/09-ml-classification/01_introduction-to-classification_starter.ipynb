# Introduction to Classification

Learn the basics of classification: finding which category something belongs to.

## Classification vs. Regression

| Question | Type | Answer |
|----------|------|--------|
| Is this email spam? | Classification | Yes or No |
| What price will this house sell for? | Regression | $435,000 |
| Will this customer churn? | Classification | Churn or Stay |
| How many units will sell next month? | Regression | 12,400 units |
| What species is this flower? | Classification | Rose, Tulip, or Daisy |

**Rule of thumb:** If the answer is a **discrete category**, it's classification. If the answer is a **continuous number**, it's regression.

## Three Types of Classification

### 1. Binary Classification
**Exactly 2 possible outputs**

Examples:
- Spam detection: Spam or Not Spam
- Medical test: Disease or Healthy
- Fraud detection: Fraudulent or Legitimate
- Customer churn: Will Churn or Will Stay

One class is "positive" (the outcome you care about), the other is "negative."

### 2. Multi-class Classification
**3 or more mutually exclusive categories (each sample belongs to EXACTLY ONE)**

Examples:
- Digit recognition: 0, 1, 2, 3, 4, 5, 6, 7, 8, or 9
- Flower species: Setosa, Versicolor, or Virginica
- Sentiment analysis: Positive, Neutral, or Negative
- Customer tier: Gold, Silver, or Bronze

The key: each sample has **exactly one** label.

### 3. Multi-label Classification
**Multiple categories possible for one sample**

Examples:
- Movie genres: A film can be Action, Comedy, AND Thriller
- Document tags: An article can be tagged "politics", "climate", AND "legislation"
- Medical diagnosis: A patient has Hypertension, Diabetes, AND Sleep Apnea
- Image labels: Photo contains a Dog, a Park, AND a Child

The key: each sample can have **zero, one, or multiple** labels.

## Decision Boundaries

A classifier learns to draw a line (in 2D) or surface (in higher dimensions) that separates classes.

```
In 2D: A line separates the classes

    Class A 
    *   *   *
    *   *
    ________ <- decision boundary
        *   *   *
        *   * Class B
```

When predicting a new point:
- If it falls on Class A side → predict Class A
- If it falls on Class B side → predict Class B

## Probability Scores vs. Hard Labels

Most classifiers don't just predict a category — they output a **probability**.

```python
# Hard label approach:
prediction = "spam"  # 0 or 1, no uncertainty

# Probability approach:
probability = 0.87   # 87% confident it's spam, 13% it's not
threshold = 0.5      # If prob >= 0.5, predict spam. Otherwise, not spam.
```

Why probabilities matter:
- You can see the model's confidence
- You can adjust the threshold based on business needs
- They enable more sophisticated evaluation metrics

## The Classification Pipeline

```
1. Data Preparation
   - Clean data
   - Handle missing values
   - Create features
        ↓
2. Train/Test Split
   - 80% train, 20% test (typically)
   - Use stratify=y for balanced splits
        ↓
3. Feature Scaling
   - StandardScaler or MinMaxScaler
        ↓
4. Train Classifier
   - Choose algorithm (e.g., logistic regression)
   - Fit on training data
        ↓
5. Predict on Test Data
   - Generate predictions
        ↓
6. Evaluate
   - Accuracy, precision, recall, F1, ROC-AUC
        ↓
7. Iterate & Improve
   - Adjust features, algorithm, threshold
```

## Key Terms

| Term | Definition |
|------|------------|
| **Class** | One of the possible output categories |
| **Label** | The true class of a sample |
| **Prediction** | What the model outputs for a sample |
| **Positive class** | The "interesting" outcome in binary classification |
| **Negative class** | The default or non-interesting outcome |
| **Stratification** | Ensuring train/test splits maintain class proportions |
| **Decision boundary** | The surface separating classes |
| **Probability threshold** | The cutoff for converting probability to a class label |
