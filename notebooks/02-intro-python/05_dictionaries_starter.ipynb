## Creating a Dictionary

A dictionary stores data as key-value pairs using curly braces.

```python
student = {
    "name": "Alex",
    "age": 25,
    "major": "Data Science",
    "gpa": 3.8
}

print(student)
print(type(student))
```

## Accessing Values

Retrieve values using square brackets or the `.get()` method.

```python
student = {"name": "Alex", "age": 25}

# Using square brackets
print(student["name"])      # Output: Alex

# Using .get() - safer method
print(student.get("age"))   # Output: 25
print(student.get("city", "Unknown"))  # Output: Unknown (default value)
```

## Modifying a Dictionary

Add, update, and delete key-value pairs.

```python
student = {"name": "Alex", "age": 25}

# Add a new key
student["major"] = "Data Science"
print(student)

# Update an existing key
student["age"] = 26
print(student)

# Delete a key-value pair
del student["major"]
print(student)
```

## Iterating Through a Dictionary

Loop through keys, values, or both.

```python
student = {"name": "Alex", "age": 25, "major": "Data Science"}

# Iterate through keys
for key in student:
    print(key)

print("---")

# Iterate through keys and values
for key, value in student.items():
    print(f"{key}: {value}")
```
