## Creating Lists

A list stores an ordered collection of items in square brackets.

```python
# A list of numbers
scores = [85, 92, 78, 90, 88]

# A list of strings
fruits = ["apple", "banana", "cherry"]

# An empty list
empty = []

print(scores)
print(fruits)
print(f"Length of scores: {len(scores)}")
```

## Accessing Items

Use index numbers to access individual items. Indices start at 0.

```python
fruits = ["apple", "banana", "cherry"]

print(fruits[0])      # Output: apple
print(fruits[1])      # Output: banana
print(fruits[-1])     # Output: cherry (last item)
print(fruits[-2])     # Output: banana (second to last)
```

## Slicing Lists

Extract a portion of a list using slicing syntax.

```python
scores = [85, 92, 78, 90, 88, 76, 95]

print(scores[1:4])     # Output: [92, 78, 90]
print(scores[:3])      # Output: [85, 92, 78]
print(scores[4:])      # Output: [88, 76, 95]
print(scores[::2])     # Output: [85, 78, 88, 95] (every other item)
```

## Modifying Lists

Update, add, and remove items from lists.

```python
fruits = ["apple", "banana", "cherry"]

# Update an item
fruits[1] = "mango"
print(fruits)

# Add to the end with append()
fruits.append("orange")
print(fruits)

# Remove an item
fruits.remove("banana")
print(fruits)

# Sort a list
numbers = [5, 2, 8, 1, 9]
numbers.sort()
print(numbers)
```
