## Defining a Class

Use the `class` keyword to define a class.

```python
class Student:
    pass   # placeholder

print(Student)
```

## The `__init__()` Constructor

Initialize object attributes when creating a new instance.

```python
class Student:
    def __init__(self, name, major, gpa):
        self.name = name
        self.major = major
        self.gpa = gpa

# Create objects (instances) of the class
student1 = Student("Alex", "Data Science", 3.8)
student2 = Student("Jordan", "Computer Science", 3.5)

print(student1.name)    # Output: Alex
print(student1.major)   # Output: Data Science
print(student2.gpa)     # Output: 3.5
```

## Methods

Methods are functions defined inside a class that describe what objects can do.

```python
class Student:
    def __init__(self, name, major, gpa):
        self.name = name
        self.major = major
        self.gpa = gpa

    def introduce(self):
        return f"Hello, I'm {self.name} and I study {self.major}."

    def is_honor_student(self):
        return self.gpa >= 3.7

# Create and use objects
student1 = Student("Alex", "Data Science", 3.8)
print(student1.introduce())
print(f"Honor student: {student1.is_honor_student()}")

student2 = Student("Jordan", "Computer Science", 3.5)
print(student2.introduce())
print(f"Honor student: {student2.is_honor_student()}")
```

## Modifying Object Attributes

Update object attributes using dot notation.

```python
class Student:
    def __init__(self, name, major, gpa):
        self.name = name
        self.major = major
        self.gpa = gpa

student = Student("Alex", "Data Science", 3.5)
print(f"GPA: {student.gpa}")

# Update the GPA
student.gpa = 3.9
print(f"Updated GPA: {student.gpa}")
```
