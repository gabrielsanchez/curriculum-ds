Run the cell below to load the student records. Do not modify this cell.

Write code to:

## Task 2: Write a Cleaning Function

Write a function `clean_student(record)` that:
- Takes a single student dict as input
- Returns a **new dict** (do not modify the original) with the `name` field cleaned: leading/trailing whitespace removed and converted to title case
- Leaves all other fields unchanged

Test your function on the first student record.

```python
def clean_student(record):
    return {
        "name": record["name"].strip().title(),
        "gpa": record["gpa"],
        "major": record["major"],
    }


# Test it
print(clean_student(students[0]))  # {'name': 'Alice Smith', 'gpa': 3.8, 'major': 'Computer Science'}
print(clean_student(students[1]))  # {'name': 'Bob Jones', 'gpa': 1.9, 'major': 'Mathematics'}
```

## Task 3: Write a Grade Function

Write a function `letter_grade(gpa)` that maps a numeric GPA to a letter grade:

| GPA range | Grade |
|-----------|-------|
| >= 3.5    | A     |
| >= 3.0    | B     |
| >= 2.0    | C     |
| >= 1.0    | D     |
| < 1.0     | F     |

Test your function by printing the letter grade for each student.

```python
def letter_grade(gpa):
    if gpa >= 3.5:
        return "A"
    elif gpa >= 3.0:
        return "B"
    elif gpa >= 2.0:
        return "C"
    elif gpa >= 1.0:
        return "D"
    else:
        return "F"


# Test it — print name and grade for each student
for student in students:
    cleaned = clean_student(student)
    grade = letter_grade(cleaned["gpa"])
    print(f"{cleaned['name']}: GPA {cleaned['gpa']} -> {grade}")

# Alice Smith: GPA 3.8 -> A
# Bob Jones: GPA 1.9 -> D
# Carol  White: GPA 2.7 -> C
# David Brown: GPA 3.2 -> B
# Eve Davis: GPA 0.9 -> F
# Frank Miller: GPA 3.5 -> A
# Grace Wilson: GPA 2.1 -> C
# Henry Moore: GPA 3.9 -> A
```

## Task 4: List Comprehensions

Using list comprehensions, create:
1. `passing_students` — a list of cleaned student dicts where GPA >= 2.0
2. `curved_gpas` — a list of GPAs for all students after applying a 5% curve, capped at 4.0 (i.e., `min(gpa * 1.05, 4.0)`)

Print both lists.

```python
# passing_students: cleaned records where gpa >= 2.0
passing_students = [clean_student(s) for s in students if s["gpa"] >= 2.0]

# curved_gpas: all gpas after 5% curve, capped at 4.0
curved_gpas = [round(min(s["gpa"] * 1.05, 4.0), 2) for s in students]

print(f"Passing students: {len(passing_students)}")  # Passing students: 6
for s in passing_students:
    print(f"  {s['name']} ({s['gpa']})")
# Alice Smith (3.8)
# Carol  White (2.7)
# David Brown (3.2)
# Frank Miller (3.5)
# Grace Wilson (2.1)
# Henry Moore (3.9)

print(f"\nCurved GPAs: {curved_gpas}")
# Curved GPAs: [3.99, 2.0, 2.84, 3.36, 0.94, 3.67, 2.2, 4.0]
```

## Task 5: Grade Summary Dictionary

Build a dictionary called `grade_summary` where each key is a letter grade ('A', 'B', 'C', 'D', 'F') and each value is the count of students who received that grade.

Include all students (not just passing ones).

```python
grade_summary = {}

for student in students:
    grade = letter_grade(student["gpa"])
    grade_summary[grade] = grade_summary.get(grade, 0) + 1

print(grade_summary)  # {'A': 3, 'D': 1, 'C': 2, 'B': 1, 'F': 1}
```

## Task 6: StudentRoster Class

Define a class `StudentRoster` with:
- `__init__(self, records)` — stores the records and applies `clean_student` to each one
- `summary_report(self)` — prints the total number of students, average GPA (rounded to 2 decimal places), and the grade distribution from `grade_summary`
- `top_student(self)` — returns the student dict with the highest GPA

Create a `StudentRoster` from the `students` list, call `summary_report()`, and print the top student.

```python
class StudentRoster:
    def __init__(self, records):
        self.records = [clean_student(r) for r in records]

    def summary_report(self):
        total = len(self.records)
        avg_gpa = round(sum(s["gpa"] for s in self.records) / total, 2)
        grades = {}
        for s in self.records:
            g = letter_grade(s["gpa"])
            grades[g] = grades.get(g, 0) + 1
        print(f"Total students:   {total}")
        print(f"Average GPA:      {avg_gpa}")
        print(f"Grade distribution: {grades}")

    def top_student(self):
        return max(self.records, key=lambda s: s["gpa"])


roster = StudentRoster(students)
roster.summary_report()
print("\nTop student:", roster.top_student())

# Total students:   8
# Average GPA:      2.88
# Grade distribution: {'A': 3, 'D': 1, 'C': 2, 'B': 1, 'F': 1}
#
# Top student: {'name': 'Henry Moore', 'gpa': 3.9, 'major': 'Data Science'}
```
