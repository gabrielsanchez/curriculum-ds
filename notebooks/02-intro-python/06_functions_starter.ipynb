## Defining and Calling a Function

Use the `def` keyword to create a function.

```python
def greet():
    print("Hello, welcome to Data Science!")

# Call the function
greet()
```

## Functions with Parameters

Parameters allow you to pass information into a function.

```python
def greet(name):
    print(f"Hello, {name}! Welcome to Data Science!")

greet("Alex")
greet("Jordan")

# Multiple parameters
def describe_student(name, age, city):
    print(f"{name} is {age} years old and lives in {city}.")

describe_student("Alex", 25, "Chicago")
```

## Return Statements

Use `return` to send a value back from a function.

```python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)  # Output: 8

def calculate_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "F"

grade = calculate_grade(85)
print(f"Grade: {grade}")
```

## Default Parameters

Provide default values for parameters.

```python
def greet_with_title(name, title="Friend"):
    print(f"Hello, {title} {name}!")

greet_with_title("Alex")
greet_with_title("Alex", "Dr.")

def multiply(a, b=2):
    return a * b

print(multiply(5))       # Output: 10 (uses default b=2)
print(multiply(5, 3))    # Output: 15
```
