## Declaring Variables

In Python, you create a variable by writing its name, an `=` sign, and the value you want to store.

```python
# Declaring variables
name = "Alex"
age = 25
height = 5.9
is_student = True

print(name)
print(age)
print(height)
print(is_student)
```

## Updating Variables

You can change the value of a variable by assigning it a new value.

```python
age = 26
print(age)
```

## Understanding Data Types

Use the `type()` function to see what data type a variable holds.

```python
# Integer (whole number)
num_students = 30
print(type(num_students))

# Float (decimal number)
temperature = 98.6
print(type(temperature))

# String (text)
city = "Chicago"
print(type(city))

# Boolean (True or False)
is_active = True
print(type(is_active))
```
