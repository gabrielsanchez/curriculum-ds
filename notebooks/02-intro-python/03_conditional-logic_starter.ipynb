## Comparison Operators

Comparison operators return `True` or `False`.

```python
score = 85
print(score == 85)    # Output: True
print(score != 90)    # Output: True
print(score > 80)     # Output: True
print(score < 80)     # Output: False
print(score >= 85)    # Output: True
print(score <= 85)    # Output: True
```

## The `if` Statement

Execute code only when a condition is `True`.

```python
temperature = 38

if temperature > 37.5:
    print("You have a fever.")
```

## `if` / `else`

Execute one block when the condition is `True`, another when it's `False`.

```python
score = 72

if score >= 60:
    print("You passed!")
else:
    print("You did not pass.")
```

## `if` / `elif` / `else`

Check multiple conditions in sequence.

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```

## Logical Operators

Combine conditions using `and`, `or`, and `not`.

```python
age = 25
has_license = True

# Using 'and'
if age >= 18 and has_license:
    print("You can drive.")

# Using 'or'
if age < 13 or age > 65:
    print("Special discount applies.")
else:
    print("Regular price.")

# Using 'not'
if not has_license:
    print("You need a license to drive.")
else:
    print("You have a valid license.")
```
