## Creating Strings

Create strings using single quotes, double quotes, or triple quotes for multi-line strings.

```python
name = "Alex"
city = 'Chicago'
bio = """Alex is a data science student
who loves working with data."""

print(name)
print(city)
print(bio)
```

## String Concatenation

Combine strings using the `+` operator or use f-strings for cleaner code.

```python
first_name = "Data"
last_name = "Science"
full_name = first_name + " " + last_name
print(full_name)

# Using f-strings
name = "Alex"
age = 25
message = f"My name is {name} and I am {age} years old."
print(message)
```

## String Indexing and Slicing

Access individual characters or extract substrings using indexing and slicing.

```python
word = "Python"

# Indexing - access individual characters
print(word[0])      # Output: P
print(word[1])      # Output: y
print(word[-1])     # Output: n (last character)

# Slicing - extract substrings
print(word[0:3])    # Output: Pyt (characters 0, 1, 2)
print(word[2:])     # Output: thon (from index 2 to end)
print(word[:4])     # Output: Pyth (start to index 3)
print(word[::2])    # Output: Pto (every 2nd character)
```

## Common String Methods

Python provides built-in methods to transform and analyze strings.

```python
text = "hello world"

# Convert to uppercase or lowercase
print(text.upper())         # Output: HELLO WORLD
print(text.lower())         # Output: hello world

# Replace part of a string
print(text.replace("world", "Python"))  # Output: hello Python

# Split a string into a list
print(text.split())         # Output: ['hello', 'world']

# Check if a substring exists
print("world" in text)      # Output: True
```
