# Keras Framework - Practice Exercise

Build and optimize neural networks on MNIST:
1. Experiment with different architectures
2. Compare optimizers (SGD, Adam, RMSprop)
3. Tune hyperparameters (batch size, learning rate)
4. Analyze training curves and convergence

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# Load MNIST
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()
X_train = X_train / 255.0
X_test = X_test / 255.0

print(f"Data loaded: {X_train.shape}, {X_test.shape}")
```

## Exercise: Compare Optimizers

Keras supports multiple optimizers. Test SGD, Adam, and RMSprop on the same architecture.

```python
# Build same model architecture
def build_model():
    return tf.keras.Sequential([
        tf.keras.layers.Flatten(input_shape=(28, 28)),
        tf.keras.layers.Dense(128, activation='relu'),
        tf.keras.layers.Dense(64, activation='relu'),
        tf.keras.layers.Dense(10, activation='softmax')
    ])

# Test different optimizers
optimizers = {
    'SGD': tf.keras.optimizers.SGD(learning_rate=0.01),
    'Adam': tf.keras.optimizers.Adam(learning_rate=0.001),
    'RMSprop': tf.keras.optimizers.RMSprop(learning_rate=0.001)
}

results = {}
for opt_name, optimizer in optimizers.items():
    print(f"\nTraining with {opt_name} optimizer...")
    model = build_model()
    model.compile(optimizer=optimizer, loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    hist = model.fit(X_train, y_train, epochs=10, batch_size=128, 
                     validation_split=0.2, verbose=0)
    test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
    results[opt_name] = {'history': hist, 'test_acc': test_acc}
    print(f"{opt_name}: Test Accuracy = {test_acc:.4f}")

# Plot comparison
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

for opt_name in optimizers.keys():
    hist = results[opt_name]['history']
    axes[0].plot(hist.history['val_accuracy'], label=opt_name, marker='o', markersize=4)

axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Validation Accuracy')
axes[0].set_title('Optimizer Comparison: Convergence Speed')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Test accuracy
names = list(results.keys())
test_accs = [results[n]['test_acc'] for n in names]
axes[1].bar(names, test_accs, alpha=0.7)
axes[1].set_ylabel('Test Accuracy')
axes[1].set_title('Final Test Accuracy by Optimizer')
axes[1].set_ylim([0.95, 1.0])
for i, v in enumerate(test_accs):
    axes[1].text(i, v + 0.002, f'{v:.4f}', ha='center', fontweight='bold')
axes[1].grid(True, axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\nQuestion: Which optimizer converges fastest? Which achieves best test accuracy?")
```
