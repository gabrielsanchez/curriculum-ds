# Summary Statistics

Learn to calculate and interpret summary statistics: mean, median, mode, and standard deviation.

```python
import pandas as pd
import numpy as np
```

## Create a housing dataset

```python
df = pd.DataFrame({
    "price": [215000, 180000, 340000, 290000, 125000, 410000, 175000, 260000, 198000, 1250000, 305000, 222000],
    "bedrooms": [3, 2, 4, 3, 2, 5, 2, 3, 3, 6, 4, 3],
    "sqft": [1400, 1100, 2100, 1800, 900, 2800, 1050, 1600, 1350, 5200, 2000, 1450],
    "age_years": [12, 25, 5, 18, 40, 3, 30, 10, 22, 8, 6, 15],
    "neighborhood": ["North", "South", "North", "East", "South", "North", "East", "South", "North", "North", "East", "South"]
})

print(df.head())
```

## Quick Overview: `.describe()`

```python
df.describe()
```

## Mean

```python
mean_price = df["price"].mean()
print(f"Mean price: ${mean_price:,.0f}")
```

## Median

```python
median_price = df["price"].median()
print(f"Median price: ${median_price:,.0f}")
print(f"Mean vs Median: {mean_price - median_price:,.0f} difference")
```

## Mode

```python
mode_bedrooms = df["bedrooms"].mode()[0]
mode_neighborhood = df["neighborhood"].mode()[0]

print(f"Most common bedrooms: {mode_bedrooms}")
print(f"Most common neighborhood: {mode_neighborhood}")
```

## Standard Deviation and Range

```python
std_price = df["price"].std()
price_range = df["price"].max() - df["price"].min()

print(f"Standard deviation: ${std_price:,.0f}")
print(f"Price range: ${price_range:,.0f}")
print(f"Min: ${df['price'].min():,.0f}, Max: ${df['price'].max():,.0f}")
```

## Percentiles and IQR

```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)
IQR = Q3 - Q1

print(f"Q1 (25th percentile): ${Q1:,.0f}")
print(f"Q3 (75th percentile): ${Q3:,.0f}")
print(f"IQR: ${IQR:,.0f}")
```
