# Correlation Matrix

Measure relationships between variables and detect multicollinearity.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Create a sample dataset

```python
np.random.seed(42)
n = 500

sqft = np.random.normal(1600, 400, n).clip(600)
bedrooms = (sqft / 400 + np.random.normal(0, 0.5, n)).clip(1, 6).round()
age_years = np.random.normal(18, 10, n).clip(0)
price = (sqft * 120 + bedrooms * 8000 - age_years * 1500 + np.random.normal(0, 20000, n)).clip(80000)
score = (70 - age_years * 0.3 + np.random.normal(0, 5, n)).clip(0, 100)
garage = np.random.choice([0, 1, 2], n, p=[0.2, 0.5, 0.3])

df = pd.DataFrame({
    "price": price,
    "sqft": sqft,
    "bedrooms": bedrooms,
    "age_years": age_years,
    "score": score,
    "garage": garage
})

print(df.head())
```

## Calculate the correlation matrix

```python
corr_matrix = df.corr()
print(corr_matrix.round(2))
```

## Interpret correlation strength

Magnitude guide:
- 0.00-0.19: Negligible
- 0.20-0.39: Weak
- 0.40-0.59: Moderate
- 0.60-0.79: Strong
- 0.80-1.00: Very strong

```python
print("Key relationships:")
print(f"Price vs sqft: {corr_matrix.loc['price', 'sqft']:.2f} (Strong positive)")
print(f"Price vs age: {corr_matrix.loc['price', 'age_years']:.2f} (Moderate negative)")
print(f"Sqft vs bedrooms: {corr_matrix.loc['sqft', 'bedrooms']:.2f} (Strong positive - multicollinearity!)")
print(f"Garage vs price: {corr_matrix.loc['garage', 'price']:.2f} (Weak)")
```

## Visualize with a heatmap

```python
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1, vmax=1,
    linewidths=0.5,
    square=True
)
plt.title("Correlation Matrix — Housing Features")
plt.tight_layout()
plt.show()
```

## Mask the upper triangle (reduce clutter)

```python
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1, vmax=1,
    linewidths=0.5,
    square=True
)
plt.title("Correlation Matrix (Lower Triangle)")
plt.tight_layout()
plt.show()
```
