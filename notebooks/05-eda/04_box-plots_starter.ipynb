# Box Plots

Use box plots to understand distributions and compare groups.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Create a sample dataset

```python
np.random.seed(42)

df = pd.DataFrame({
    "price": np.concatenate([
        np.random.normal(220000, 35000, 300),
        np.random.normal(280000, 45000, 300),
        np.random.normal(195000, 28000, 300),
    ]),
    "neighborhood": ["South"] * 300 + ["North"] * 300 + ["East"] * 300,
    "bedrooms": np.random.choice([1, 2, 3, 4, 5], 900, p=[0.05, 0.25, 0.40, 0.20, 0.10])
})

df.loc[0, "price"] = 950000
df.loc[1, "price"] = 875000
```

## Basic box plot

```python
plt.figure(figsize=(6, 6))
plt.boxplot(df["price"], patch_artist=True,
            boxprops=dict(facecolor="steelblue", color="navy"),
            medianprops=dict(color="white", linewidth=2))
plt.title("Distribution of House Prices")
plt.ylabel("Price ($)")
plt.xticks([1], ["All Homes"])
plt.tight_layout()
plt.show()
```

## Grouped box plots

```python
groups = [df[df["neighborhood"] == n]["price"] for n in ["South", "North", "East"]]

plt.figure(figsize=(8, 6))
bp = plt.boxplot(groups,
                 labels=["South", "North", "East"],
                 patch_artist=True,
                 boxprops=dict(facecolor="steelblue", alpha=0.7),
                 medianprops=dict(color="red", linewidth=2))
plt.title("House Price Distribution by Neighborhood")
plt.xlabel("Neighborhood")
plt.ylabel("Price ($)")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## Box plots with Seaborn (easier for grouped data)

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x="neighborhood", y="price", palette="Set2", order=["South", "North", "East"])
plt.title("House Price Distribution by Neighborhood")
plt.xlabel("Neighborhood")
plt.ylabel("Price ($)")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## Multiple box plots for different columns

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].boxplot(df["bedrooms"], patch_artist=True, boxprops=dict(facecolor="coral", alpha=0.7))
axes[0].set_title("Number of Bedrooms")
axes[0].set_ylabel("Bedrooms")

axes[1].boxplot(df["price"], patch_artist=True, boxprops=dict(facecolor="steelblue", alpha=0.7))
axes[1].set_title("House Price")
axes[1].set_ylabel("Price ($)")

plt.suptitle("Comparing Distributions")
plt.tight_layout()
plt.show()
```
