# Introduction to EDA

This notebook introduces the EDA workflow and demonstrates the standard first steps when encountering a new dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Create a sample dataset

```python
df = pd.DataFrame({
    "price": [215000, 180000, 340000, 290000, 125000, 410000, 175000, 260000, 198000, 1250000, 305000, 222000],
    "bedrooms": [3, 2, 4, 3, 2, 5, 2, 3, 3, 6, 4, 3],
    "sqft": [1400, 1100, 2100, 1800, 900, 2800, 1050, 1600, 1350, 5200, 2000, 1450],
    "age_years": [12, 25, 5, 18, 40, 3, 30, 10, 22, 8, 6, 15],
    "neighborhood": ["North", "South", "North", "East", "South", "North", "East", "South", "North", "North", "East", "South"]
})
```

## Step 1: Check shape and size

```python
print(f"Dataset shape: {df.shape}")
print(f"Dimensions: {df.shape[0]} rows, {df.shape[1]} columns")
```

## Step 2: Check data types

```python
df.info()
```

## Step 3: Check for missing values

```python
print("Missing values per column:")
print(df.isnull().sum())
```

## Step 4: Check for duplicates

```python
print(f"Duplicate rows: {df.duplicated().sum()}")
```

## Step 5: Quick summary statistics
